# Maxima locaux function

```python
def find_local_maxima(s: np.ndarray):
    """
    Find local maxima indices.
    """

    s = np.asarray(s)

    return np.where(
        (s[1:-1] > s[:-2]) &
        (s[1:-1] > s[2:])
    )[0] + 1
```

# Minima locaux function

```python
def find_local_minima(s: np.ndarray):
    """
    Find local minima indices.
    """

    s = np.asarray(s)

    return np.where(
        (s[1:-1] < s[:-2]) &
        (s[1:-1] < s[2:])
    )[0] + 1
```
